```python
import pandas as pd

# Read CSV files from drive
matches = pd.read_csv('/content/drive/MyDrive/T20_Predictor_Project/matches.csv')
deliveries = pd.read_csv('/content/drive/MyDrive/T20_Predictor_Project/deliveries.csv')

# Checking data is loaded or not
print("Matches Data Preview:")
display(matches.head(3))

print("\nDeliveries Data Preview:")
display(deliveries.head(3))
```

```python
# SQL Logic
total_data = deliveries.merge(matches, left_on='match_id', right_on='match_id')

print("Data merge successfully!")
print("Shape of the new merged dataset (Rows, Columns):", total_data.shape)
display(total_data.head(3))
```

```python
# Checking for missing (null) values in our dataset
print("Missing values in each column:")
print(total_data.isnull().sum())
```

```python
# Dropping rows where the match has no winner(No Result Matches)
total_data.dropna(subset=['winner'], inplace=True)

# Filling missing values in wicket-related columns with '0'
total_data['dismissed_player'] = total_data['dismissed_player'].fillna('0')
total_data['dismissal_type'] = total_data['dismissal_type'].fillna('0')
total_data['fielder'] = total_data['fielder'].fillna('0')
total_data['extra_type'] = total_data['extra_type'].fillna('0')

print("Data cleaned successfully!")
print("Checking if any missing values are left in 'winner':", total_data['winner'].isnull().sum())
print("New data shape after cleaning:", total_data.shape)
```

```python
# Checking for missing (null) values in our dataset
print("Missing values in each column:")
print(total_data.isnull().sum())
```

```python
# Filter data to get only the 1st innings
first_innings = total_data[total_data['innings'] == 1]

# Calculate total runs scored in the 1st innings for each match
total_scores = first_innings.groupby('match_id')['total_runs'].sum().reset_index()

# Add 1 to get the target score for the chasing team
total_scores['target'] = total_scores['total_runs'] + 1

# Merge this target score back with our main dataset
total_data = total_data.merge(total_scores[['match_id', 'target']], on='match_id')

# Keep only the 2nd innings data (since our model predicts the chase)
chasing_data = total_data[total_data['innings'] == 2]

print("Target scores calculated successfully!")
print("Shape of our new Chasing Data:", chasing_data.shape)
display(chasing_data.head(3))
```

```python
# Creating a copy of the data to avoid Pandas warnings
chasing_data = chasing_data.copy()

# Calculating the 'current_score' by taking the cumulative sum of runs per ball
chasing_data['current_score'] = chasing_data.groupby('match_id')['total_runs'].cumsum()

# Calculating 'runs_left' by subtracting the current score from the target
chasing_data['runs_left'] = chasing_data['target'] - chasing_data['current_score']

# Displaying important columns to verify the new calculations
print("Current Score and Runs Left calculated successfully!")
display(chasing_data[['match_id', 'target', 'total_runs', 'current_score', 'runs_left']].head(5))
```

```python
# Calculating 'balls_left'
# In this dataset, 'over' starts at 1. So total balls bowled = (over - 1) * 6 + ball
balls_bowled = (chasing_data['over'] - 1) * 6 + chasing_data['ball']
chasing_data['balls_left'] = 120 - balls_bowled

# Ensuring 'balls_left' doesn't go below 0 (which can happen due to extra balls in the 20th over)
chasing_data['balls_left'] = chasing_data['balls_left'].apply(lambda x: 0 if x < 0 else x)

# Calculating 'wickets_left'
# Taking the cumulative sum of the 'is_wicket' column to count fallen wickets
wickets_fallen = chasing_data.groupby('match_id')['is_wicket'].cumsum()
chasing_data['wickets_left'] = 10 - wickets_fallen

# Displaying the final critical columns
print("Balls Left ans Wickets Left calculated successfully!")
display(chasing_data[['match_id', 'runs_left', 'balls_left', 'wickets_left']].head(5))
```

```python
# Calculating Current Run Rate (CRR)
# Formula: (Current Score / Balls Bowled) * 6
# Balls bowled can be found by subtracting balls_left from total 120 balls
balls_bowled = 120 - chasing_data['balls_left']
chasing_data['crr'] = (chasing_data['current_score'] * 6) / balls_bowled

# Calculating Required Run Rate (RRR)
# Formula: (Runs Left / Balls Left) * 6
chasing_data['rrr'] = (chasing_data['runs_left'] * 6) / chasing_data['balls_left']

# Displaying the Run Rated to verify
print("CRR and RRR calculated successfully!")
display(chasing_data[['match_id', 'current_score', 'runs_left', 'crr', 'rrr']].head(5))
```
